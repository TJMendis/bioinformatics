```python
class IndexHash(object):
    def __init__(self,t,k):
        self.k = k
        self.index = {}

        for i in range(len(t)-k+1):
            kmer = t[i:i+k]
            
            if kmer in self.index:
                self.index[kmer].append(i)
            else:
                self.index[kmer]=[i]
            
    def query(self,p):
        kmer = p[:self.k]
        return self.index.get(kmer, [])
        """
        if kmer in self.index:
            return self.index[kmer]
        else:
            return []
        """    

def query_index(t,p,idx):
    k=idx.k
    matches = []
    offsets = idx.query(p)
    for i in offsets:
        if t[i+k:i+len(p)]==p[k:]:
            matches.append(i)
    return matches
    
```

```python
t = 'ACTTGGAGATCTTTGAGGCTAGGTATTCGGGATCGAAGCTCATTTCGGGGATCGATTACGATATGGTGGGTATTCGGGA'
p = 'GGTATTCGGGA'
idx = IndexHash(t, 4)
print(query_index(t, p, idx))    # expect [21, 68]
```
