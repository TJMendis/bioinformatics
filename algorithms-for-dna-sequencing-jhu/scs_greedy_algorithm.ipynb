```python
from itertools import permutations
```

```python
def overlap(a, b, min_length=1):
    start = 0  # start all the way at the left
    while True:
        start = a.find(b[:min_length], start)  # look for b's prefix in a
        if start == -1:  # no more occurrences to right
            return 0
        # found occurrence; check for full suffix/prefix match
        if b.startswith(a[start:]):
            return len(a) - start
        start += 1  # move just past previous match


def find_biggest_overlap(reads):
    max_overlap_score = 0
    l_best_read = ""
    r_best_read = ""
    for read_1, read_2 in permutations(reads, 2):
        overlap_score = overlap(read_1, read_2)
        if overlap_score > max_overlap_score:
            max_overlap_score = overlap_score
            l_best_read = read_1
            r_best_read = read_2
    return max_overlap_score, l_best_read, r_best_read


def greedy_scs(reads):
    reads = list(reads)
    score, left, right = find_biggest_overlap(reads)
    while score > 0:
        reads.remove(left)
        reads.remove(right)
        reads.append(left + right[score:])
        score, left, right = find_biggest_overlap(reads)
    return ''.join(reads)
```

```python
reads = ["ABC", "DEF", "GHJ"]
print(greedy_scs(reads))
```
