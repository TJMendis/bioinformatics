```python
import bisect
class IndexSorted:
    def __init__(self,t,k):
        self.k = k
        self.index = []
        for i in range(len(t)-k+1):
            self.index.append((t[i:i+k],i))
        self.index.sort()
    def query(self, p):
        kmer = p[:self.k]
        i =bisect.bisect_left(self.index,(kmer,0))
        hits = []
        while i < len(self.index):
            if self.index[i][0]!=kmer:
                break
            
            hits.append(self.index[i][1])
            i+=1
        return hits

def query_index(p,t,index):
    k=index.k
    matches = []
    for i in index.query(p):
        if t[i+k:i+len(p)]==p[k:]:
            matches.append(i)
    return matches
            
    
```
