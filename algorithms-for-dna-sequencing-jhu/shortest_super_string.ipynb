```python
from itertools import permutations
```

```python
def overlap(a, b, min_length=1):
    start = 0  # start all the way at the left
    while True:
        start = a.find(b[:min_length], start)  # look for b's prefix in a
        if start == -1:  # no more occurrences to right
            return 0
        # found occurrence; check for full suffix/prefix match
        if b.startswith(a[start:]):
            return len(a) - start
        start += 1  # move just past previous match


def string_builder(read_1, read_2):
    overlap_score = overlap(read_1, read_2)
    if overlap_score > 0:
        return read_1 + read_2[overlap_score:]
    else:
        return read_1 + read_2


def shortest_string_prob(reads):
    ss_list = []
    permutation_stream = itertools.permutations(reads)
    for perm in permutation_stream:
        cum_read = ""
        for i in range(len(perm)):
            cum_read = string_builder(cum_read, perm[i])
        ss_list.append(cum_read)
    return ss_list
```

```python
reads = ["AAT", "ATG", "GCT"]
print(shortest_string_prob(reads))
print(min(shortest_string_prob(reads)))
reads_2 = ["ATG", "TGC", "CAT"]
print(shortest_string_prob(reads_2))
print(min(shortest_string_prob(reads_2)))
print(min(shortest_string_prob(reads_2), key=len))
```
