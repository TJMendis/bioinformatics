```python
class Graph:
    def __init__(self):
        self.nodes = {}

    def add_node(self, kmer):
        if kmer[:-1] not in self.nodes:
            node = Node(kmer[:-1])
            node.add_out_edge(kmer)
            self.nodes[kmer[:-1]] = node
        else:
            self.nodes[kmer[:-1]].add_out_edge(kmer)
        if kmer[1:] not in self.nodes:
            node = Node(kmer[1:])
            node.add_in_edge(kmer)
            self.nodes[kmer[1:]] = node
        else:
            self.nodes[kmer[1:]].add_in_edge(kmer)

    def find_start(self):
        for node in self.nodes.values():
            if not node.in_edge:
                return node.kmer

    def seq_construction(self):
        node_name = self.find_start()
        seq = node_name
        while self.nodes[node_name].out_edge:
            seq += self.nodes[node_name].out_edge[0][-1]
            node_name = self.nodes[node_name].out_edge.pop(0)[1:]

        return seq


class Node:
    def __init__(self, kmer):
        self.kmer = kmer
        self.out_edge = []
        self.in_edge = []

    def add_out_edge(self, kmer):
        if self.kmer == kmer[:-1]:
            self.out_edge.append(kmer)

    def add_in_edge(self, kmer):
        if self.kmer == kmer[1:]:
            self.in_edge.append(kmer)
```

```python
def make_kmers(reads, k):
    kmers = []
    for read in reads:
        for i in range(len(read) - k + 1):
            kmers.append(read[i : i + k])
    return kmers
```

```python
# Langmed test case
# here this works on sheer luck due to the order (fifo is happening)
reads = ["a_long_long_long_time"]
reads_2 = ["a_long_long_long", "ng_long_l", "g_long_time"]
k = 8
graph_3 = Graph()
for kmer in make_kmers(reads, k):
    graph_3.add_node(kmer)

print(graph_3.seq_construction())
```
