```python
def overlap_imp(a, b, min_length):
    if a.find(b[:min_length]) == -1:
        return 0
    start = 0
    while True:
        if start > len(a) - min_length:
            return 0
        start = a.find(b[:min_length], start)
        if start == -1:
            return 0
        if b.startswith(a[start:]):
            return len(a) - start

        start += 1
```

```python
def overlap_graph(min_length, reads):
    overlap_struct = {}
    for i in range(len(reads)):
        for j in range(len(reads)):
            if i == j:
                continue
            overlap_score = overlap_imp(reads[i], reads[j], min_length)
            if overlap_score > 0:
                overlap_struct[(reads[i], reads[j])] = overlap_score
    return overlap_struct
```

```python
reads = ["TTACGA", "CGATCC", "TCCAAT"]
print(overlap_graph(3, reads))
```

```python
# Langmead test case
reads = ["ACGGATGATC", "GATCAAGT", "TTCACGGA"]
print(overlap_graph(3, reads))
```
