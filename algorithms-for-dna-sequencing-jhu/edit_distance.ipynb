```python
def edit_distance(x, y):
    if len(x) == 0:
        return len(y)
    if len(y) == 0:
        return len(x)
    ed_1 = edit_distance(x[:-1], y[:-1]) + delta(x[-1], y[-1])
    ed_2 = edit_distance(x, y[:-1]) + 1
    ed_3 = edit_distance(x[:-1], y) + 1
    return min(ed_1, ed_2, ed_3)


def delta(x, y):
    if x == y:
        return 0
    else:
        return 1
```

```python
import time

t0 = time.time()
print(edit_distance("Shakespeare", "shake spear"))
print(f"{time.time() - t0:.2f}s")
```
