```python
def delta(x, y):
    if x == y:
        return 0
    else:
        return 1
```

```python
memo = {}


def edit_distance_dp(a, b):

    if (a, b) in memo:
        return memo[(a, b)]
    if len(a) == 0:
        return len(b)
    if len(b) == 0:
        return len(a)
    ed_1 = edit_distance_dp(a[:-1], b[:-1]) + delta(a[-1], b[-1])
    ed_2 = edit_distance_dp(a, b[:-1]) + 1
    ed_3 = edit_distance_dp(a[:-1], b) + 1

    memo[(a, b)] = min(ed_1, ed_2, ed_3)
    return memo[(a, b)]
```

```python
def edit_distance_dpm(a, b):
    matrix = [[0] * (len(b) + 1) for _ in range(len(a) + 1)]
    for j in range(len(b) + 1):
        matrix[0][j] = j
    for i in range(len(a) + 1):
        matrix[i][0] = i
    for i in range(1, len(a) + 1):
        for j in range(1, len(b) + 1):
            # here we move from matrix convention to python string convention
            ed_1 = matrix[i - 1][j - 1] + delta(a[i - 1], b[j - 1])
            ed_2 = matrix[i][j - 1] + 1
            ed_3 = matrix[i - 1][j] + 1
            matrix[i][j] = min(ed_1, ed_2, ed_3)
    return matrix[len(a)][len(b)]
```

```python
test_cases = [
    ("ABC", "BBC", 1),
    ("Shakespeare", "shake spear", 3),
    ("kitten", "sitting", 3),  # classic Levenshtein example
    ("intention", "execution", 5),  # another classic
    ("", "ABC", 3),  # edge: empty first string
    ("ABC", "", 3),  # edge: empty second string
    ("", "", 0),  # edge: both empty
]

memo.clear()
for a, b, expected in test_cases:
    dp_result = edit_distance_dp(a, b)
    dpm_result = edit_distance_dpm(a, b)
    status = "✓" if (dp_result == dpm_result == expected) else "✗"
    print(
        f"{status} edit_distance({a!r}, {b!r}): top-down={dp_result}, bottom-up={dpm_result}, expected={expected}"
    )
```
